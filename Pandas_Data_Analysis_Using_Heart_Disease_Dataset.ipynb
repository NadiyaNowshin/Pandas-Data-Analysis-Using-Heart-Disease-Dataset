Task 3: Categorical Data Analysis

```python
#Count patients with and without heart disease

print("Heart disease:", (df["target"] == 1).sum())
print("No heart disease:" , (df["target"] == 0).sum())
```

```python
#Count male and female patients

print("No of male:", (df["sex"] == 1).sum())
print("No of female:" , (df["sex"] == 0).sum())
```

Task 4: Data Selection Using iloc

```python
#Select the first 10 rows

print(df.iloc[:10])
```

```python
#Extract age, sex, and cholesterol

print(df.iloc[: , [0,1,4]])
```

```python
#Select rows from index 20 to 30 and columns from index 0 to 4

print(df.iloc[20:31, 0:5])
```

Task 5: Data Filtering Using Conditions

```python
#Extract all patients older than 50 years

patients_over_50 = df[df["age"] > 50]

print(patients_over_50)
```

```python
#Extract patients who have heart disease and a cholesterol level greater than 240

patients = df[(df["target"] == 1) & (df["chol"] > 240)]

print(patients)
```

```python
#Count how many patients satisfy both conditions

count = df[(df["target"] == 1) & (df["chol"] > 240)].shape[0]

print(count)
```
